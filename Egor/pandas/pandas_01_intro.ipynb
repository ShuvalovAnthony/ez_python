## 2. Как обращаться к ячейкам

- столбец по имени: `df["балл"]`;
- ячейка по номеру строки: `df["балл"].iloc[pos]`;
- ячейка по метке индекса: `df.loc[label, "балл"]`;
- срез `iloc` не включает правый конец, срез `loc` включает;
- один фильтр: `df.loc[df["балл"] > 3, "балл"]`.

```python
scores = pd.DataFrame({"балл": [3, 1, 6, 6, 2]})
print(scores["балл"].iloc[1])
print(scores.loc[3, "балл"])
print(scores["балл"].iloc[1:3].tolist())
print(scores["балл"].loc[1:3].tolist())
print(scores.loc[scores["балл"] > 3, "балл"].tolist())
```

Верни кортеж:

1. ячейка `df[col].iloc[pos]`;
2. ячейка `df.loc[label, col]`;
3. список значений `df[col].iloc[start:end]`;
4. список значений `df[col].loc[start:end]`;
5. список значений `col`, которые строго больше `threshold`.
